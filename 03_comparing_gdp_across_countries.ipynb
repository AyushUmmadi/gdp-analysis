# GDP Analysis — Part 3: Comparing GDP Across Countries

Individual charts are useful, but the real value comes from putting countries **side by side** on the same chart — this notebook compares raw GDP levels between selected countries and against the world total, using color-coded interactive line charts.

(Growth-*rate* comparisons — a different, more nuanced question — are covered separately in Part 4.)

```python
import os
import pandas as pd
import plotly.express as px
import plotly.offline as pyo

df = pd.read_csv('data/gdp.csv')
os.makedirs('output', exist_ok=True)
```

## All Countries at Once — A Broad Overview

Plotting every country on a single chart is intentionally cluttered — the point isn't to read individual lines, but to get a sense of overall scale and spread before narrowing down to specific comparisons.

```python
fig = px.line(df, x='Year', y='Value', title='GDP Analysis of All Countries', color='Country Name')
pyo.plot(fig, filename='output/all_countries_gdp.html', auto_open=False)
```

## India vs. China

```python
china = df[df['Country Name'] == 'China']
india = df[df['Country Name'] == 'India']
compare_df = pd.concat([china, india], axis=0)

fig = px.line(compare_df, x='Year', y='Value', title='GDP Comparison | India & China', color='Country Name')
pyo.plot(fig, filename='output/india_china_gdp.html', auto_open=False)
fig
```

## China vs. World Total

```python
china = df[df['Country Name'] == 'China']
world = df[df['Country Name'] == 'World']
compare_df = pd.concat([china, world], axis=0)

fig = px.line(compare_df, x='Year', y='Value', title='GDP Comparison | World & China', color='Country Name')
pyo.plot(fig, filename='output/world_china_gdp.html', auto_open=False)
fig
```

**Observation:** China's GDP has grown dramatically relative to the world total in recent decades, closing much of the gap between the two lines — a trend that's far easier to see once the two are overlaid on the same chart.
