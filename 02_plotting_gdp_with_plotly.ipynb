# GDP Analysis — Part 2: Plotting GDP with Plotly

Plotly is used here instead of static plotting libraries because it produces **interactive** charts — hover to see exact values, zoom into a time range, and export standalone HTML files that work offline, without needing Python installed to view them.

This notebook:
1. Plots World and India's GDP as interactive line charts
2. Automates the same plot for **every** country in the dataset, saving each as its own HTML file

```python
import os
import pandas as pd
import plotly.express as px
import plotly.offline as pyo

df = pd.read_csv('data/gdp.csv')
df.head()
```

## 1. Single-Country Plots

```python
df_world = df[df['Country Name'] == 'World']

fig = px.line(df_world, x='Year', y='Value', title='World GDP Analysis')
fig
```

```python
os.makedirs('output', exist_ok=True)
pyo.plot(fig, filename='output/World_GDP.html', auto_open=False)
```

```python
df_india = df[df['Country Name'] == 'India']

fig = px.line(df_india, x='Year', y='Value', title='India GDP Analysis')
fig
```

```python
pyo.plot(fig, filename='output/India_GDP.html', auto_open=False)
```

## 2. Plotting Graphs in Bulk

Automate the same chart for every country using a loop, saving each as a separately named HTML file — 256 charts generated in one pass.

```python
os.makedirs('output/individual_gdp', exist_ok=True)

for country_name in df['Country Name'].unique():
    country_df = df[df['Country Name'] == country_name]
    fig = px.line(country_df, x='Year', y='Value', title=country_name + ' GDP Analysis')
    pyo.plot(fig, filename=f'output/individual_gdp/{country_name}.html', auto_open=False)

print('Generated', df['Country Name'].nunique(), 'individual GDP charts.')
```

Each country now has its own interactive HTML chart under `output/individual_gdp/`, viewable offline in any browser.
